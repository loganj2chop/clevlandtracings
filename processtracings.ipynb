```python
import pandas as pd
```

```python
df = pd.read_csv("clevelandvudssofar.csv")
```

```python
df = df[df['Recivedtxtfiles'].notna()]
```

```python
df = df[df['age'].notna()].reset_index()
```

```python
df.head()
```

```python
df.columns = df.columns.str.strip()
df['avg_rating'] = ((df['review1'] + df['review2'] + df['review3']) / 3).round(0).astype(int)
df = df.drop(columns=['Study'])
```

```python
df["study_id"] = df["Textfile"]
```

```python
df.to_csv("clevelandmani.csv", index=False)
```
